### Constructor Theory (`__init__` Method)
* Functions inside a class are known as **methods**.
* `__init__` is a special method (Dunder method / Double Underscore method).
* It runs **automatically** when you create an object from a class.
* Used for the **initialization of the object's attributes**, Set initial values to the object attributes.
* A class can have **only one** `__init__` method.

```python
# Example of dunder method
class Car:
    def __init__(self):
        print("in init method")
        
    def fun1(self):
        print("in fun1")

obj = Car() # Object created; __init__ method is executed automatically
```

#### Understanding `self`
* `self` carries the object reference.
* Reference to the current object (instance).
* Used to initialize and access attributes and methods.
* When we create `obj1`, `obj2`, `obj3`, etc., they take the place of the `self` parameter.
* **`self` is not a keyword.** You can use any word (e.g., `inext`, `me`, `args`), but `self` is standard per PEP 8 and Python documentation.

```python
# self example
class Company:
    def __init__(self, i, n, c):
        self.id = i
        self.name = n
        self.city = c

    def show(self):
        print("ID:", self.id)
        print("Name:", self.name)
        print("City:", self.city)

obj1 = Company(101, "Keshav", "Pune")
obj2 = Company(102, "Rahul", "Mumbai")

obj1.show()
obj2.show()
```

## Types of Variables in a Class

1. **Instance Variable**
2. **Class Variable**
3. **Local Variable**

###1.1 Instance Variable
* Defined inside `__init__` (or any instance method) using `self`.
* The value of an instance variable varies from object to object (Object-level variable).
* Multiple copies are created (object to object).
* Always access using `self.variable_name` inside the class.
* Stores data unique to each object.
* You can access instance varibale in any method using self keyword:
### Instance Method 
* Oprates on object(instance) data
* Can access and modify instance variable
* Its having 'self' as the first parameter

```python
#1.2 Example: Instance Variable & Instance Method
class Student:
    def __init__(self, name, roll):
        self.name = name  # Instance variable
        self.roll = roll  # Instance variable

    def show(self):       # Instance method operates on object data
        print(f"Name: {self.name}, Roll: {self.roll}")

s1 = Student("Sourabh", 101)
s2 = Student("Bob", 102)

s1.show()  # Output: Name: Sourabh, Roll: 101
s2.show()  # Output: Name: Bob, Roll: 102
```

### 2.1 Class Variable
* Operates on class-level data.Belongs to the class
* Only one varibale will be created and shared across all the objects
* Declared inside the class , outside of any methods
* access using `ClassName.variable_name`.

#### Class Method
* Its operates on class variable , Class level data 
* Decorated with `@classmethod`.
* Takes the class itself as the first parameter, conventionally named `cls`.

```python
#2.2 Example: Class Variable & Class Method

class Student:
    school_name = "ARC School"  # Class variable

    @classmethod
    def change_school(cls, new_name):
        cls.school_name = new_name

Student.change_school("Prashant Vidyalaya")
print(Student.school_name)  # Output: Prashant Vidyalaya
```

### 3.1 Local Variable
* Defined inside a method.
* Stores values for temporary purposes.
* Scope is restricted **within the method only**.

```python
# 3.2 Local variable example
class Test:
    def display(self):
        msg = "Hello"  # Local variable
        print(msg)

t = Test()
t.display()  # Output: Hello
# print(msg)  # Error: name 'msg' is not defined outside
```

###3.3 Static Method
* Acts like a normal function inside a class.
* Doesn't need access to instance (`self`) or class (`cls`).
* Decorated with `@staticmethod`.
* Useful for utility or helper functions inside a class.

```python
#3.4 Static method example
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(5, 3))  # Output: 8
```
